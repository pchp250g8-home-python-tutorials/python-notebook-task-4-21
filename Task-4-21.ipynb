В двух бригадах более 27 человек. Число членов 1-ой бригады, более чем в 2 раза превышает число членов
2-ой бригады, уменьшенное на 12. Число членов 2-ой бригады, более чем в 9 раз превышает число членов 1-ой
бригады, уменьшенное на 10. Сколько человек во 2-ой бригаде?

```python
from sympy.abc import x,y
from sympy import solve

ineq1 = x + y > 27
ineq2 = x > 2 * (y - 12)
ineq3 = y > 9 * (x - 10)
ineq4 = x > 0
ineq5 = y > 0
ineq1a = solve(ineq1, x)
ineq1b = solve(ineq1, y)
ineq2a = ineq2
ineq2b = solve(ineq2, y)
ineq3a = solve(ineq3, x)
ineq3b = ineq3
ineq4a = ineq4
ineq4b = ineq5
sys_ineq1 = [(ineq1a.lhs - ineq2a.lhs)  >  (ineq1a.rhs - ineq2a.rhs),
            (ineq3a.lhs - ineq4a.lhs)  <  (ineq3a.rhs - ineq4a.rhs)]
sys_ineq2 = [(ineq1b.lhs - ineq2b.lhs)  <  (ineq1b.rhs - ineq2b.rhs),
            (ineq3b.lhs - ineq4b.lhs)  >  (ineq3b.rhs - ineq4b.rhs)]
print(sys_ineq1)
print(sys_ineq2)
print(solve(sys_ineq1,x))
print(solve(sys_ineq2,y))
```
