В двух бригадах более 27 человек. Число членов 1-ой бригады, более чем в 2 раза превышает число членов
2-ой бригады, уменьшенное на 12. Число членов 2-ой бригады, более чем в 9 раз превышает число членов 1-ой
бригады, уменьшенное на 10. Сколько человек во 2-ой бригаде?

```python
from sympy.abc import x,y
from sympy import solve
from sympy import sympify
from sympy import reduce_inequalities

ineq1 = x + y > 27
ineq2 = x > 2 * (y - 12)
ineq3 = y > 9 * (x - 10)
#--исключение переменной y
sys_ineq1 = reduce_inequalities([ineq1,ineq2], y) 
ineq11 = sys_ineq1.args[0] #Выразили переменную y из 1-го неравенства
ineq12 = sys_ineq1.args[1] #Выразили переменную y из 2-го неравенства
ineq11 = (ineq11.gts - ineq12.lts > ineq11.lts - ineq12.gts).simplify()
ineq12 = (ineq12.gts - ineq3.lts > ineq12.lts - ineq3.gts).simplify()
#--исключение переменной x
sys_ineq2 = reduce_inequalities([ineq1,ineq3], x)
ineq21 = sys_ineq2.args[0] #Выразили переменную x из 1-го неравенства
ineq22 = sys_ineq2.args[1] #Выразили переменную x из 3-го неравенства
ineq21 = (ineq21.gts - ineq22.lts > ineq21.lts - ineq22.gts).simplify()
ineq22 = (ineq22.gts - ineq2.lts > ineq22.lts - ineq2.gts).simplify()
print(solve([ineq11,ineq12],x))
print(solve([ineq21,ineq22],y))

```
